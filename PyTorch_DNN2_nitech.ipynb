```python
import numpy as np
import torch
from torch import nn, optim
import matplotlib.pyplot as plt
import pandas as pd
import sklearn
from sklearn.model_selection import train_test_split
from tqdm.notebook import tqdm

```

# 1. 屈折率データ(sample数3000件超)を用いて演習

```python
path=str("./oxide_refractive_index.no_outliers.csv")
df_original=pd.read_csv(path, index_col=0)

df = df_original.dropna()

drop_columns = ['refractive_index','composition']
df_descriptor = df.drop(drop_columns, axis=1)
target_columns = ['refractive_index']
df_target = df[target_columns]

display(df_descriptor)
display(df_target)
```

■　保留法　test_size=0.3 として、 X, t -->  train_X, test_X, train_t, test_t  を作成せよ

```python
X=df_descriptor.values
t=df_target.values

train_X, test_X, train_t, test_t = train_test_split(X,t,test_size=0.3)
print (len(train_X), len(test_X))
```

■　numpy -->  Tensor形式変換  (train_X --> train_X のまま)

```python
train_X= torch.Tensor(train_X)
train_t= torch.Tensor(train_t)
test_X= torch.Tensor(test_X)
test_t= torch.Tensor(test_t)

train_X.shape
```

■　modelの定義  1784の記述子を徐々に 1つのノードに集約させること。

```python
model= nn.Sequential(
    nn.Linear(201,128),
    nn.Sigmoid(),
    nn.Linear(128,16),
    nn.Sigmoid(),
    nn.Linear(16,1),
)
```

■　Loss関数の定義。 MSE (mean square error)

```python
loss_fn = nn.MSELoss()   #Loss関数の定義　この場合は MSE
```

```python
#optimizer = optim.SGD(model.parameters(), lr=0.001)  #model.parameters() 学習させたい変数　　ln は学習率
optimizer = optim.Adam(model.parameters(), lr=0.001)  #model.parameters() 学習させたい変数　　ln は学習率
```

■　DNNの実行

```python
model.train()   # modelをtraining モードにする
loss_history=[]      #  lossのステップごとの推移
losstest_history=[]
for epoch in range(200):
    optimizer.zero_grad()   #optimizerの初期化
    train_y=model(train_X)   #train_y と train_tは異なることに注意
    test_y=model(test_X)
    loss=loss_fn(train_y,train_t)   #Loss関数の計算
    losstest=loss_fn(test_y,test_t)
    loss_history.append(float(loss))
    losstest_history.append(float(losstest))
    loss.backward()   #傾きを計算
    optimizer.step()   #更新処理実施

plt.plot(loss_history)  #blue
plt.plot(losstest_history) #orange
```

```python
print (loss_fn(train_y,train_t))
plt.plot(train_y.detach().numpy(),train_t.detach().numpy(),'x')
test_y=model(test_X)
plt.plot(test_y.detach().numpy(),test_t.detach().numpy(),'+')
```

Modelの再定義

BatchNorm1d() 数値の規格化処理を実施 　回帰分析の結果をかなり良くすることがある。  （いつも有効とは限らないことに注意）<BR>
ReLU()  活性化関数処理<BR>

```python
p=0.0

model= nn.Sequential(
    nn.Linear(201,128),
    nn.BatchNorm1d(128),
    nn.ReLU(),
    nn.Dropout(p),
    nn.Linear(128,64),
    nn.BatchNorm1d(64),
    nn.ReLU(),
    nn.Dropout(p),
    nn.Linear(64,16),
    nn.BatchNorm1d(16),
    nn.ReLU(),
    nn.Linear(16,1),
)
```

Minibatchの利用 loader

```python
from torch.utils.data import TensorDataset, DataLoader

dataset = TensorDataset(train_X, train_t)
loader = DataLoader(dataset,batch_size=64, shuffle=True)
```

```python
#optimizer = optim.SGD(model.parameters(), lr=0.001)  #model.parameters() 学習させたい変数　　ln は学習率
optimizer = optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-3)  #model.parameters() 学習させたい変数　　ln は学習率
```

```python
loss_history=[]
losstest_history=[]

for epoch in tqdm(range(200)):
    model.train()
    for X, t in loader:
        optimizer.zero_grad()   #optimizerの初期化
        y=model(X)
        loss=loss_fn(y,t)
        loss.backward()   #傾きを計算
        optimizer.step()   #更新処理実施

    loss_history.append(float(loss))
    test_y=model(test_X)
    losstest=loss_fn(test_y,test_t)
    losstest_history.append(float(losstest))



plt.plot(loss_history)  #blue
plt.plot(losstest_history) #orange
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, r2_score

model.eval()  # 評価モード（BatchNorm, Dropoutを固定）

with torch.no_grad():
    train_y = model(train_X)
    test_y = model(test_X)

# NumPyに変換
train_pred = train_y.cpu().numpy().ravel()
train_true = train_t.cpu().numpy().ravel()

test_pred = test_y.cpu().numpy().ravel()
test_true = test_t.cpu().numpy().ravel()

# RMSE, R2
train_rmse = np.sqrt(mean_squared_error(train_true, train_pred))
train_r2 = r2_score(train_true, train_pred)

test_rmse = np.sqrt(mean_squared_error(test_true, test_pred))
test_r2 = r2_score(test_true, test_pred)

print(f"Train: RMSE = {train_rmse:.4f}, R2 = {train_r2:.4f}")
print(f"Test : RMSE = {test_rmse:.4f}, R2 = {test_r2:.4f}")

# プロット
plt.figure(figsize=(6, 6))

plt.plot(train_pred, train_true, '+',
         label=f'Train (RMSE={train_rmse:.3f}, R²={train_r2:.3f})')

plt.plot(test_pred, test_true, 'x',
         label=f'Test (RMSE={test_rmse:.3f}, R²={test_r2:.3f})')

# y=x の理想線
vmin = min(train_true.min(), test_true.min(),
           train_pred.min(), test_pred.min())
vmax = max(train_true.max(), test_true.max(),
           train_pred.max(), test_pred.max())

plt.plot([vmin, vmax], [vmin, vmax], 'k--')

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
